#Aufgabe 11: Products - Formate von price und promo_price erkennen

```python
# Direkt für price und promo_price ausgeben:
for col in ['price', 'promo_price']:
    print(f"Format-Übersicht für: {col}")
    print(products_df[col].astype(str).str.replace(r'\d', '9', regex=True).value_counts())
    print("\n")
```

#Aufgabe 12: Products - Umwandeln von Formaten:
* von   9.999.999 in   9999.999
* von  99.999.999 in  99999.999
* von 999.999.999 in 999999.999

```python
# Für die 'price'-Spalte:
zwei_punkte_price = products_df['price'].str.count(r'\.') >= 2

products_df.loc[zwei_punkte_price, 'price'] = (products_df.loc[zwei_punkte_price, 'price'].str.replace(".", "", n=1, regex=False))


# Für die 'promo_price'-Spalte:
zwei_punkte_promo = products_df['promo_price'].str.count(r'\.') >= 2

products_df.loc[zwei_punkte_promo, 'promo_price'] = (products_df.loc[zwei_punkte_promo, 'promo_price'].str.replace(".", "", n=1, regex=False))
```

#Aufgabe 13: Products - Formaten wieder anzeigen lassen

```python
for col in ['price', 'promo_price']:
    print(f"Format-Übersicht für: {col}")
    print(products_df[col].astype(str).str.replace(r'\d', '9', regex=True).value_counts())
    print("\n")
```

#Aufgabe 14: Products - Änderung der Datentypen (inklusiv type-Spalte) und Vergleich zwischen price und promo_price
* Falls  price >= promo_price -> Keine Änderung in promo_price.
* Falls  price < promo_price -> Promo_price = 0

```python
# 1. Datentypen für price und promo_price umwandeln
products_df['price'] = products_df['price'].astype(float)
products_df['promo_price'] = products_df['promo_price'].astype(float)

# 2. Spalte 'type' bereinigen (Komma zu Punkt) und in Ganzzahlen (int) umwandeln
products_df['type'] = products_df['type'].astype(str).str.replace(',', '.', regex=False).astype(float).astype(int)

# 3. Vergleich: Falls price < promo_price, wird promo_price auf 0 gesetzt
ungueltig = products_df['price'] < products_df['promo_price']
products_df.loc[ungueltig, 'promo_price'] = 0.0
```

# 1. DATEN LADEN

```python
import re
produkte = pd.read_csv('products.csv')
print(f"Original: {produkte.shape}")
```

```python
def ist_kaputter_preis(wert):
    s = str(wert)
    if s.count('.') > 1:  # z.B. 1.119.976
        return True
    if re.search(r'\.\d{3}$', s):  # z.B. 499.899
        return True
    return False
```

# 2. DUPLIKATE ENTFERNEN

```python
produkte_sauber = produkte.drop_duplicates(subset=['sku']).copy()
print(f"Nach Duplikate entfernen: {produkte_sauber.shape}")
```

# 3. KAPUTTE PREISE LÖSCHEN

```python
maske_kaputt_price = produkte_sauber['price'].astype(str).apply(ist_kaputter_preis)
print(f"Kaputte Katalogpreise: {maske_kaputt_price.sum()}")

produkte_sauber = produkte_sauber[~maske_kaputt_price].copy()
produkte_sauber = produkte_sauber.drop(columns=['promo_price'])
print(f"Nach Price-Cleaning: {produkte_sauber.shape}")
```

# 4. FEHLENDE WERTE FÜLLEN

```python
produkte_sauber['desc'] = produkte_sauber['desc'].replace(['[]', '[ ]', ''], pd.NA).fillna('Keine Beschreibung')
produkte_sauber['type'] = produkte_sauber['type'].replace(['[]', '[ ]', ''], pd.NA)
produkte_sauber['type'] = produkte_sauber['type'].fillna(produkte_sauber['type'].mode()[0])
```

# 5. DATENTYPEN KORRIGIEREN

```python
produkte_sauber['price'] = pd.to_numeric(produkte_sauber['price'], errors='coerce')
produkte_sauber['sku'] = produkte_sauber['sku'].astype(str)
produkte_sauber['name'] = produkte_sauber['name'].astype(str)
produkte_sauber['in_stock'] = produkte_sauber['in_stock'].astype(int)
```

# 6. Null-Preise löschen

```python
produkte_sauber = produkte_sauber.dropna(subset=['price'])

print(f"\nFinal: {produkte_sauber.shape}")
produkte_sauber.info()
```

# 7. CHECK: Wie viele Macs/iPhones behalten wir?

```python
macs = produkte_sauber[produkte_sauber['name'].str.contains('MacBook|iPhone|iMac', case=False, na=False)]
print(f"\nMacs/iPhones erhalten: {len(macs)} (bei 667 wären es nur 293)")
```
